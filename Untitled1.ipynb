Feature Engineering.
we create new features in this part.



Lets start by creating aggregated features eg total_minutes, total_calls, total_charge: Help summarize customer usage.


```python
# Create new features
df["total_minutes"] = df["total day minutes"] + df["total eve minutes"] + df["total night minutes"] + df["total intl minutes"]
df["total_calls"] = df["total day calls"] + df["total eve calls"] + df["total night calls"] + df["total intl calls"]
df["total_charge"] = df["total day charge"] + df["total eve charge"] + df["total night charge"] + df["total intl charge"]
```

Then we create Interaction Features, i.e 

avg_call_duration: Helps understand if short or long calls affect churn.
charge_per_minute: High charge per minute may indicate dissatisfaction.

```python
# Interaction Features
df["avg_call_duration"] = df["total_minutes"] / df["total_calls"]
df["charge_per_minute"] = df["total_charge"] / df["total_minutes"]
```

lastly we create Binary Flags:

high_service_calls: High service calls may indicate unhappy customers.
high_day_minutes: High usage customers might be at risk of leaving.
high_intl_usage: Customers with high international usage might have different churn patterns.

```python
# Binary Features (Flags)
df["high_service_calls"] = (df["customer service calls"] > df["customer service calls"].median()).astype(int)
df["high_day_minutes"] = (df["total day minutes"] > df["total day minutes"].median()).astype(int)
df["high_intl_usage"] = (df["total intl minutes"] > df["total intl minutes"].median()).astype(int)
```

Let's run the code and maybe later try to visualise our new features!

```python
# Handle infinite values (from division)
df.replace([np.inf, -np.inf], np.nan, inplace=True)

# Fill NaNs if any
df.fillna(0, inplace=True)

# Save the transformed dataset
df.to_csv("transformed_dataset.csv", index=False)

print("Feature engineering completed! New features added.")
```

HANDLING OUTLIERS IN THE TOTAL MINUTES & CHARGE_PER_MINUTE.

Instead of removing outliers, capping them is better (Winsorization) so we don’t lose data.

```python


df["total_minutes"] = winsorize(df["total_minutes"], limits=[0.01, 0.01])  # Capping at 1% on both sides
df["charge_per_minute"] = winsorize(df["charge_per_minute"], limits=[0.01, 0.01])
```

Lets perfom hyperparameter tuning!

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression

# Define hyperparameter grid
param_grid = {
    'C': [0.01, 0.1, 1, 10],  # Regularization strength
    'penalty': ['l1', 'l2'],  # Lasso (L1) or Ridge (L2)
    'solver': ['liblinear']    # Best solver for L1 & L2
}

# Initialize Logistic Regression model
log_reg = LogisticRegression(max_iter=1000, random_state=42)

# Perform GridSearchCV
grid_search = GridSearchCV(log_reg, param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train, y_train)

# Get best hyperparameters
best_params = grid_search.best_params_
print("Best Hyperparameters:", best_params)

# Train model with best parameters
best_log_reg = LogisticRegression(**best_params, max_iter=1000, random_state=42)
best_log_reg.fit(X_train, y_train)

# Predict using best model
y_pred_best = best_log_reg.predict(X_test)
```

```python
# Evaluate the model
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("\nTuned Model Accuracy:", accuracy_score(y_test, y_pred_best))
print("\nClassification Report:\n", classification_report(y_test, y_pred_best))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV

# Define parameter grid for tuning
param_grid = {
    'n_estimators': [100, 200, 300, 500],  # Number of trees
    'max_depth': [10, 20, 30, None],  # Depth of trees
    'min_samples_split': [2, 5, 10],  # Minimum samples for split
    'min_samples_leaf': [1, 2, 4],  # Minimum samples per leaf
    'bootstrap': [True, False]  # Use bootstrap sampling
}

# Initialize Random Forest model
rf_model = RandomForestClassifier(random_state=42)

# Initialize RandomizedSearchCV
rf_random = RandomizedSearchCV(estimator=rf_model, 
                               param_distributions=param_grid, 
                               n_iter=20, 
                               cv=5, 
                               verbose=2, 
                               random_state=42, 
                               n_jobs=-1)

# Fit the model
rf_random.fit(X_train, y_train)

# Get the best parameters
print("Best Hyperparameters:", rf_random.best_params_)

# Evaluate on test set
best_rf = rf_random.best_estimator_
y_pred = best_rf.predict(X_test)

# Print updated accuracy
from sklearn.metrics import accuracy_score, classification_report
print("\nTuned Model Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```
